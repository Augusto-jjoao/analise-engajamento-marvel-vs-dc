## 1. Contexto e Objetivos
Este trabalho investiga o comportamento de fãs das franquias Marvel e DC no YouTube (2020-2025). O objetivo não é definir qual marca é superior, mas responder a duas perguntas centrais através de Processamento de Linguagem Natural (NLP):
1. As comunidades possuem comportamentos sociais diferentes ou são os mesmos consumidores genéricos?
2. Como cada base de fãs reage a fatores externos (polêmicas reais e marketing)?

## 2. Metodologia e Ferramentas
Utilizamos a API do YouTube para extração massiva de comentários de trailers. Para a análise semântica, adotamos o **BERTopic**, um algoritmo avançado de NLP que utiliza:
* **SBERT (Embeddings):** Para compreender o contexto e agrupar comentários por significado, não apenas por palavras exatas.
* **c-TF-IDF:** Para identificar as palavras-chave exclusivas de cada tópico, penalizando termos genéricos (como "filme" ou "trailer").
* **NLTK e Scikit-Learn:** Para limpeza de dados (Stopwords) e vetorização.

```python
# Importação das bibliotecas principais utilizadas no projeto
import json
import nltk
from bertopic import BERTopic
from sklearn.feature_extraction.text import CountVectorizer
from nltk.corpus import stopwords
import matplotlib.pyplot as plt
from wordcloud import WordCloud

nltk.download('stopwords')
```

## 3. Resultados: O Comportamento da Marvel (Comportamento de Manada)
A análise dos dados da Marvel revelou um comportamento focado no **Macro** e pautado por tendências da Geração Z.

* **Foco no Universo (Macro):** O público se preocupa mais com a continuidade e o planejamento corporativo ("Fase 4", "Canon", "MCU") do que com o filme isolado.
* **"Descaso" Com os Atores:** Diferente da DC, não há muitos tópicos sobre os atores, e quando há, é mais sobre os seus trabalhos passados do que sobre o trabalho daquele filme específico (ex: Kit Harrington no filme "Eternos" sendo mencionado nos comentários por conta do seu personagem Jon Snow).
* **Linguagem de Meme e Tópicos Rasos:** Alta incidência de gírias virais ("Sus", "Amogus") e consumo rápido guiado por *timestamps*. A maior parte dos tópicos grandes é focada nisso, o que gerou um aspecto de que os comentários da Marvel são muito mais rasos que os da DC.
* **Blindagem de Marca:** O universo cinematográfico protege o produto de polêmicas reais. No caso do ator Jonathan Majors, os tópicos focam massivamente em elogios à performance e ao vilão (Kang), ignorando questões jurídicas pessoais.

```python
# Exemplo de carregamento do modelo Marvel e visualização de tópicos focados no Macro
topic_model_marvel = BERTopic.load("modelo_marvel_final")

# Visualização de barras para o Tópico 38 (Foco na Fase 4 / Universo)
fig_marvel = topic_model_marvel.visualize_barchart(topics=[38], n_words=10, title="Marvel: Foco no Macro")
fig_marvel.show()
```

## 4. Resultados: O Comportamento da DC (Foco nos filmes individualmente)
A análise da DC revelou um comportamento focado no **Micro**, com discussões mais politizadas e combativas.

* **Foco na Obra (Micro):** O engajamento é profundo e focado no tom do filme, estética e atuação (ex: discussões sobre "Batman Fascista", comparações de atuação ou comentários sobre a cor do filme). Os tópicos maiores eram sobre os atores de cada filme analisado.
* **Rivalidade e "Papeadas":** A comunidade usa a área de comentários como arena de debate, focando em refutar críticos e comparar qualidades técnicas, muitas vezes pautando a Marvel como régua de comparação, além de comentários dos fãs da Marvel criticando os personagens da DC, o que não foi visto nos comentários da outra franquia.
* **As Consequências Externas:** Diferente da Marvel, a marca DC é vulnerável à vida real dos atores. Polêmicas envolvendo Ezra Miller e Amber Heard dominaram tópicos inteiros com vocabulário jurídico ("Jail", "Court", "Fire", "Boicott"), contaminando a percepção da obra, já que no caso da DC, os atores são o produto, e não o aspecto geral da marca.

```python
# Exemplo de carregamento do modelo DC e visualização do impacto de polêmicas reais
topic_model_dc = BERTopic.load("modelo_dc_final")

# Visualização de barras para tópicos de Tribunal (ex: Tópico 70 - Ezra Miller)
fig_dc = topic_model_dc.visualize_barchart(topics=[21], n_words=10, title="DC: Consequência de Fatores Externos")
fig_dc.show()
```

## 5. Semelhanças Entre as Duas Marcas

* **Trailers Mostram Demais:** Ambas as franquias reclamam dos seus trailers mostrarem cenas cruciais do filme, que deviam ser guardadas para o filme em si.
* **Saturação do Gênero:** Ambas as franquias também mostraram estarem cansadas do gênero de herói. Essa análise pode mostrar insatisfação dos fãs, mas também de forma geral, onde os não-fãs de Marvel e DC podem estar insatisfeitos com a quantidade de filmes e o hype em cima do gênero.

**Abaixo estão os códigos que foram usados para gerar os gráficos de tópicos da Marvel e da DC, podendo alterar o título e a cor das barras de acordo com a franquia.**

```python
import json
import nltk
from sklearn.feature_extraction.text import CountVectorizer
from bertopic import BERTopic
from nltk.corpus import stopwords

ESTUDIO = "Marvel"                          # Escolha: "Marvel" ou "DC"
NUMERO_DO_TOPICO = 20                       # Coloque o número do tópico que deseja visualizar
TITULO_DO_GRAFICO = "Tópico 20: Fórmula Repetitiva (Generic/Same)" 
NOME_DO_ARQUIVO = "Grafico_Personalizado_Topico_20.png"

# Palavras que você NÃO quer que apareçam no gráfico (lixo genérico):
PALAVRAS_PARA_REMOVER = [
    "movie", "film", "trailer", "video", "marvel", "dc", "mcu", "dceu", "studio",
    "wb", "disney", "scene", "people", "guy", "one", "watch", "watching", 
    "make", "made", "know", "think", "look", "looks", "love", "like", "best",
    "good", "great", "amazing", "hype", "wait", "comment"
]

# --- CONFIGURAÇÃO DE LIMPEZA ---
nltk.download('stopwords', quiet=True)
stops_padrao = list(set(stopwords.words('portuguese') + stopwords.words('english') + stopwords.words('spanish')))
final_stops = stops_padrao + PALAVRAS_PARA_REMOVER
vectorizer_personalizado = CountVectorizer(stop_words=final_stops)

# --- LÓGICA DE GERAÇÃO (Não precisa alterar) ---
print(f"--- Gerando Gráfico para {ESTUDIO} (Tópico {NUMERO_DO_TOPICO}) ---")

# Define as cores e arquivos baseados no estúdio
if ESTUDIO.upper() == "MARVEL":
    cor_grafico = '#E23636' # Vermelho
    arquivo_json = "dataset_Marvel.json"
    arquivo_modelo = "modelo_marvel_final"
else:
    cor_grafico = '#012269' # Azul
    arquivo_json = "dataset_DC.json"
    arquivo_modelo = "modelo_dc_final"

# Carrega os dados
with open(arquivo_json, "r", encoding="utf-8") as f:
    data_estudio = json.load(f)
textos = [c.get("text", "") for m in data_estudio.values() for c in m.get("comments", []) if len(c.get("text", "")) > 10]

# Carrega e atualiza o modelo
topic_model = BERTopic.load(arquivo_modelo)
topic_model.update_topics(textos, vectorizer_model=vectorizer_personalizado)

# Gera e salva o gráfico
try:
    fig = topic_model.visualize_barchart(
        topics=[NUMERO_DO_TOPICO], 
        n_words=10, 
        title=f"{ESTUDIO} - {TITULO_DO_GRAFICO}"
    )
    fig.update_traces(marker_color=cor_grafico)
    fig.update_layout(title_font_color=cor_grafico)
    
    # Exibe no Notebook
    fig.show()
    
    # Salva a imagem
    fig.write_image(NOME_DO_ARQUIVO)
    print(f"Sucesso! Gráfico salvo como: {NOME_DO_ARQUIVO}")
    
except Exception as e:
    print(f"Erro ao gerar o gráfico: {e}")
```
